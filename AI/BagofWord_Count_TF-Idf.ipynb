```python
import numpy as np 
import pandas as pd 
import nltk 
from nltk import word_tokenize 
from collections import Counter
```

```python
review_1="The movie was good and we really like it"
review_2="The movie was good but the ending was boring"
review_3="we did not like the movie as it was too lenghty"
review_4="I would love to see a different set of actor in the movie.A Diaster!"
```

```python
nltk.download("all")
```

```python
review_1token=word_tokenize(review_1)
review_2token=word_tokenize(review_2)
review_3token=word_tokenize(review_3)
review_4token=word_tokenize(review_4)
```

```python
review_token=set(review_1token).union(set(review_2token)).union(set(review_3)).union(review_4)
```

```python
len(review_token)
```

```python
review1_dict=Counter(review_1token)
review1_dict
```

```python
review2_dict=Counter(review_2token)
review3_dict=Counter(review_3token)
review4_dict=Counter(review_4token)
```

```python
review_dict_df=pd.DataFrame([review1_dict,review2_dict,review3_dict,review4_dict])
```

```python
review_dict_df=review_dict_df.fillna(0)
review_dict_df
```

```python
from sklearn.feature_extraction.text import CountVectorizer,TfidfVectorizer
```

```python
countvect=CountVectorizer()
```

```python
reviewlist=[review_1,review_2,review_3,review_4]
X_counts=countvect.fit_transform(reviewlist)
```

```python
X_name=countvect.get_feature_names_out()
X_name
```

```python
cv=pd.DataFrame(X_counts.toarray(),columns=X_name)
cv
```

```python
tfid=TfidfVectorizer(min_df=1,lowercase=True,stop_words='english',)
tf_Count=tfid.fit_transform(reviewlist)
tf_Count
```

```python
tfdf=pd.DataFrame(tf_Count.toarray(),columns=tfid.get_feature_names_out())
tfdf
```
