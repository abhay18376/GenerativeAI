```python
from nltk.corpus import stopwords
import re
```

```python
stpwords=stopwords.words('english')
print(stpwords)
print(stpwords)
```

```python
punctuation=re.compile(r'[-?!,:;()|0-9]',re.UNICODE)
```

```python
punt=[]
for word in token:
    puntword=punctuation.sub('',word)
    if len(puntword)>0: 
        punt.append(puntword)
```

```python
len(punt)
```

```python
punt2=[]
for word in punt:
    word=word.lower()
    if word not in stpwords: 
        punt2.append(word)
```

```python
len(punt2)
```

```python
frq=FreqDist(punt2)
```

```python
frq
```

```python
frq.most_common(10)
```

```python
import docx
```

```python
doc=docx.Document("Data/Brexit.docx")
```

```python
full_txt = []
for para in doc.paragraphs:
    if para.text.strip():
        full_txt.append(para.text)

text = " ".join(full_txt)   # convert list → string
text = ' '.join(text.split())
#clean_text = text.replace('\xa0', ' ')
print(text)
```

```python
import en_core_web_sm
npl = en_core_web_sm.load()
```

```python
import pandas as pd
import spacy
```

```python
#nlp=spacy.load('en_core_web_sm') for co llab
posdata=[]
doc1=npl(text)
for word in doc1: 
    posdata.append({
        'Token' : word.text,
        'Lemma' : word.lemma_,
        'Pos'   : word.pos_,
        'Tag'   : word.tag_, 
        'Dependency' : word.dep_
    })
df=pd.DataFrame(posdata)
df
```

```python
enn_Data=[]
for ent in doc1.ents: 
    enn_Data.append({
        'Entity':ent.text,
        'Type':ent.label_, 
        'Label':ent.label,
        'start char': ent.start_char, 
        'end char'  :  ent.end_char
    })
```

```python
entdf=pd.DataFrame(enn_Data)
```
