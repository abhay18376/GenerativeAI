```python
import nltk 
import os 
import nltk.corpus
from nltk.tokenize import word_tokenize 
from nltk.tokenize import RegexpTokenizer
```

```python
from nltk.data import load
```

```python
sent="Mary is driving a big car"
```

```python
set_token=word_tokenize(sent)
```

```python
for word in set_token: 
    print(nltk.pos_tag([word]))
```

```python
sent2="john is eating a delicious cake > an banana"
```

```python
sent2_token=word_tokenize(sent2)
```

```python
for word in sent2_token: 
    print(nltk.pos_tag([word]))
```

```python
reg_toknizer=RegexpTokenizer('(?u)\W+|\$[\d\.]+|\S+')
```

```python
result=reg_toknizer.tokenize(sent2)
result
```

```python
reg_tag=nltk.pos_tag(result)
reg_tag
```

```python
nltk.corpus.gutenberg.fileids()
```

```python
hamlet=nltk.corpus.gutenberg.words('shakespeare-hamlet.txt')
```

```python
len(hamlet)
```

```python
hamlet_pos=[]
```

```python
for word in hamlet[:2000]: 
    hamlet_pos.append(nltk.pos_tag([word]))
```

```python
hamlet_pos
```

```python
hamplet_nn=[]
for each_pos in hamlet_pos: 
    if each_pos[0][1]=="NNP": 
        hamplet_nn.append(each_pos[0][0])
```

```python
hamplet_nn
```

```python
from nltk import ne_chunk
```

```python
NE_sent="The US President stay in white House"
```

```python
NE_token=word_tokenize(NE_sent)
```

```python
NE_tag=nltk.pos_tag(NE_token)
```

```python
NE_NER=ne_chunk(NE_tag)
```
