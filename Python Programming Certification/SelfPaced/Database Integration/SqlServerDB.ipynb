```python
database=mysql.connector.connect(host="localhost",user="root",passwd="root1234",database="PythonTraining")
cursor=database.cursor()
cursor.execute("CREATE TABLE Student(name NVARCHAR(255),age SMALLINT UNSIGNED, personID int PRIMARY KEY AUTO_INCREMENT)")
cursor.execute("SHOW TABLES")
tables=cursor.fetchall()
for table in tables: 
    print(table)
```

```python
query="INSERT INTO Student (name , age) VALUES (%s , %s)"
values=('George',21)
cursor.execute(query,values)
database.commit()
print('Inserted the Record Sucessfully')
```
