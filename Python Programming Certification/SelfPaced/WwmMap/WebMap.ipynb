```python
# Single Marker

import folium
m=folium.Map()
Marker=folium.Marker(location=[12.98, 77.69], popup='Hi I am Marker', icon=folium.Icon(color='green'))
Marker.add_to(m)
m.save('Marker_Demo.html')
```

```python
import folium
m=folium.Map()
#  co ordinate with label
coordinates=[{'loc':[12.98, 77.69],'label':'Banglore'},{'loc':[12.30, 76.64],'label':'Mysore'}]
for coordinate in coordinates: 
    marker=folium.Marker(location=coordinate['loc'],popup=coordinate['label'],icon=folium.Icon(color='red'))
    marker.add_to(m)
m.save("Multiple_Marker.html")
```

```python
import pandas as pd 
import folium

df=pd.read_csv('dublin.csv')
df.head()
df1=df[["Latitude","Longitude","Name"]]
map=folium.Map(location=[df['Latitude'].mean(),df['Longitude'].mean()],zoom_start=14,control_scale=True)

m=folium.Map()
for index,location in df1.iterrows(): 
    marker=folium.Marker([location['Latitude'],location['Longitude']],popup=location['Name'],icon=folium.Icon(color='blue'))
    marker.add_to(m)
m.save('M1.html')
```
