```python
import requests
from bs4 import BeautifulSoup 
```

```python
headers = {
    'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/114.0.0.0 Safari/537.36'
}

response = requests.get('https://www.goodreads.com', headers=headers)
print(response.status_code)
```

```python
txt_categories=[]
book_soup=BeautifulSoup(response.content,"html.parser")
categories=book_soup.find_all('a',attrs={'class':'gr-hyperlink'})
for tag in categories:
    if( 'genres' in tag.get('href')):
        txt_categories.append(tag.string)
for category in txt_categories:
    print(category)
```
