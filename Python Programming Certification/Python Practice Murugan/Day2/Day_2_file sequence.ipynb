```python
# Conditional Statement
# conditional statement are used to execute statement or group of statement on certain condition
num=15
if num % 2==0:
    print('True')
    print('The Number Entered is Even')
elif num >= 10 and num !=0 :
    print('number is 10')
else:
    print('False')
print('Program completed')
```

```python
# Loop is used to execute the statement multiple time
# 1. while 2. for 3. Nested
for i in range(1,5):
    print(i)

for j in range(5,1 ,-1):
    print(j)
```

```python
# loop control statement
# 1. break 2. continue 3. pass
for i in range(1,10):
    if(i==5):
        break
    print(i)

for i in range(1,10):
    if(i==5):
        continue
    print(i)

```

```python
# Structural pattern matching
day =0
match day:
    case 1:
        print('Sunday')
    case 2:
        print('Monday')
    case _:
        print('Enter Valid Day')
    
```
