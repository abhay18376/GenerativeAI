```python
# Constructor
class fruit:
    def __init__(self,color,size):
        self.name='fruit'
        self.color=color
        self.size=size
```

```python
orange=fruit('orange',8)
print('Name of Fruit ', orange.name)
print('Color of fruit ', orange.color)
print('Size of Fruit ', orange.size)
```

```python
class Employee:
    def __init__(self,name,position):
        self.name=name
        self.position=position

    def selfinformation(self):
        print(f'My Name is {self.name} and My Position is {self.position}')

obj=Employee('Amir Khan','Manager')
obj.selfinformation()
obj2=Employee('Salman Khan','Manager')
obj2.selfinformation()
```
