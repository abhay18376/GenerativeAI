```python
# return statement is used to terminate the function and return control to   calling function
def Calculator(num1,num2):
    sum=num1+num2
    mul=num1*num2
    sub=num1-num2
    return sum,sub,mul

result =Calculator(11,22)
print(result)
```

```python
# Lambda Function
sqrt_val=lambda x : x**2

sqrt_val(10)

sum_var=lambda x: x + 22
sum_var(10)
```

```python
# Map, Filter , Reduce using Lambda
lst=[1,3,6,7]
squared=list(map(lambda x : x+2 ,lst))
print(squared)

mulvar=list(map(lambda a : a*2,lst))
print(mulvar)
```

```python
# Ffilter
fltr=list(filter(lambda x: x>4,lst))
print(fltr)
```

```python
flr=list(filter(lambda x: x>0,lst))
print(flr)
```

```python
# Reduce
import functools 
product=functools.reduce(lambda x,y : x + y ,[11,22,33,44])

print(product)
```
