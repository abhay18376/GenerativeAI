```python
class User :
    def __init__(self,name ,age ,gender):
        self.name=name
        self.age=age
        self.gender=gender

    def show_details(self):
        print('------------------------')
        print('--- personal details ---')
        print('------------------------')
        print(f'Name is : {self.name}')
        print(f'Age is : {self.age}')
        print(f'Gender is {self.gender}')
        
```

```python
class Bank(User):  #Inheritance
    def __init__(self,name,age,gender):
        super().__init__(name,age,gender) # Caonslling Base Class Constructor using super
        self.balance=0
    
    def deposit(self,amount):
        self.amount=amount
        self.balance=self.balance + self.amount
        print("Account Balance Updated , Balance Available is : ", self.balance)

    def withdraw(self,amount):
        self.amount=amount
        if(self.amount> self.balance):
            print("InSufficient Fund ...")
        else :
            self.balance=self.balance - self.amount
        print(f'Account is Debited with amount {self.amount} your working Balance is {self.balance}') 

    def ViewBalance(self):
        self.show_details()
        print('Accoiunt Balance is ',self.balance)

bank1=Bank('Amit',25,'Male')
bank1.deposit(2000)
bank1.ViewBalance()
```

```python
balance = 10000  # Starting balance
password = "1234"  # Default password

def withdraw(amount):
    global balance
    if amount > balance:
        print("Insufficient balance!")
    else:
        balance -= amount
        print(f"Withdrawal successful. New balance: ₹{balance}")

def credit(amount):
    global balance
    balance += amount
    print(f"Amount credited successfully. New balance: ₹{balance}")

def change_password():
    global password
    old_pass = getpass.getpass("Enter current password: ")
    if old_pass == password:
        new_pass = getpass.getpass("Enter new password: ")
        confirm_pass = getpass.getpass("Confirm new password: ")
        if new_pass == confirm_pass:
            password = new_pass
            print("Password changed successfully.")
        else:
            print("Passwords do not match.")
    else:
        print("Incorrect current password.")

def main():
    global password
    print("=== Welcome to XYZ Bank System ===")
    
    entered_password = getpass.getpass("Enter your password to login: ")
    if entered_password != password:
        print("Incorrect password. Access denied.")
        return

    while True:
        print("\nSelect an option:")
        print("1. Cash Withdrawal")
        print("2. Cash Credit (Deposit)")
        print("3. Change Password")
        print("4. Exit")

        choice = input("Enter your choice (1-4): ")

        if choice == '1':
            amount = float(input("Enter amount to withdraw: ₹"))
            withdraw(amount)
        elif choice == '2':
            amount = float(input("Enter amount to deposit: ₹"))
            credit(amount)
        elif choice == '3':
            change_password()
        elif choice == '4':
            print("Thank you for using XYZ Bank. Goodbye!")
            break
        else:
            print("Invalid choice. Please try again.")

# Run the program
main()

```
