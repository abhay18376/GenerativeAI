```python
from tkinter import *
from tkinter.ttk import *
```

```python
# 1
root = Tk()
root.title("Add Button")
root.geometry("300x300")

def on_button_click():
    print("Button is Clicked")

button=Button(root,text="Click Me",command=on_button_click)
button.pack(pady=50)
root.mainloop()
```

```python
# 2

root=Tk()
root.title("Spinbox Demo")
root.geometry("300x300")

spinbox=Spinbox(root,from_=0,to=10)
spinbox.pack(pady=20)
root.mainloop()
```
