```python
#1

import pandas as pd
import matplotlib.pyplot as plt

# Load the dataset (adjust the file path to where the dataset is stored on your system)
file_path = "Hurricanes.csv"  # Replace with the correct path
data = pd.read_csv(file_path)  # If it's an Excel file, use pd.read_excel(file_path)

# Check the first few rows to understand the structure of the dataset
print(data.head())

# Assuming the dataset has 'Year' and 'Number of Hurricanes' columns
# You may need to adjust the column names based on the dataset structure
years = data['Year']
hurricanes = data['Hurricanes']

# Plotting the bar graph
plt.figure(figsize=(10, 6))
plt.bar(years, hurricanes, color='blue')

# Adding titles and labels
plt.title("Number of Hurricanes in the Atlantic by Year", fontsize=16)
plt.xlabel("Year", fontsize=12)
plt.ylabel("Number of Hurricanes", fontsize=12)

# Show the plot
plt.xticks(rotation=45)  # Rotate the x-axis labels for better readability
plt.tight_layout()  # Adjust the layout to make sure everything fits
plt.show()
```

```python
#2
import pandas as pd
import matplotlib.pyplot as plt

# Load the dataset
file_path = "CityTemps.csv"
df = pd.read_csv(file_path)

# Display the first few rows to understand the structure
print(df.head())

df1=pd.DataFrame(df)

# Extract temperature data for San Francisco and Moscow
sf_temps = df1["San Francisco"]
moscow_temps = df1["Moscow"]

plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.hist(sf_temps, bins=15, color='blue', alpha=0.7, edgecolor='black')
plt.title("Temperature Distribution in San Francisco (2014-2015)")
plt.xlabel("Temperature (°C)")
plt.ylabel("Frequency")

plt.subplot(1, 2, 2)
plt.hist(moscow_temps, bins=15, color='red', alpha=0.7, edgecolor='black')
plt.title("Temperature Distribution in Moscow (2014-2015)")
plt.xlabel("Temperature (°C)")
plt.ylabel("Frequency")

plt.tight_layout()
plt.show()




```

```python
#3 

import pandas as pd
import matplotlib.pyplot as plt

# Load the dataset
file_path = "sample-salesv2.csv"
df = pd.read_csv(file_path)

# Display the first few rows to understand the structure
df.head()
#df.groupby("name")["Item Code"].nunique()
manufacturer_model_counts = df.groupby("name")["item_code"].nunique()

# Find the manufacturer with the highest number of releases
print(df.head())

top_manufacturer = manufacturer_model_counts.idxmax()
top_count = manufacturer_model_counts.max()

# Plot pie chart
plt.figure(figsize=(10, 6))
plt.pie(manufacturer_model_counts, labels=manufacturer_model_counts.index, autopct='%1.1f%%', startangle=140)
plt.title("Number of Models Released by Each Manufacturer")

# Show the chart
plt.show()

# Output the manufacturer with the most releases
top_manufacturer, top_count
```

```python
#5 

import pandas as pd
import matplotlib.pyplot as plt
# Sample Sales Data
data = {
    "name": ["Alice", "Bob", "Alice", "David", "Bob", "Alice", "David"],
    "product": ["Laptop", "Phone", "Tablet", "Monitor", "Laptop", "Phone", "Tablet"],
    "unit_price": [1000, 500, 300, 200, 1000, 500, 300],
    "quantity": [1, 2, 3, 1, 1, 2, 2],
    "net_price": [1000, 1000, 900, 200, 1000, 1000, 600],
    "date": pd.to_datetime(["2024-03-01", "2024-03-02", "2024-03-03", 
                             "2024-03-04", "2024-03-05", "2024-03-06", "2024-03-07"])
}

# Create a DataFrame
df = pd.DataFrame(data)

# Save to CSV
csv_filename = "sample-salesv2.csv"
df.to_csv(csv_filename, index=False)

# Read CSV File
df_read = pd.read_csv(csv_filename)

# Phase 2: Describe Data (Unit Price)
unit_price_description = df_read["unit_price"].describe()

# Phase 3: Filter Data and Group by Name
df_filtered = df_read[["name", "net_price", "date"]]
df_grouped = df_filtered.groupby("name").sum()

# Phase 4: Plot Total Sales by Customer
plt.figure(figsize=(8, 5))
df_grouped["net_price"].plot(kind="bar", color="skyblue")
plt.xlabel("Customer Name")
plt.ylabel("Total Sales")
plt.title("Total Sales by Customer")
plt.xticks(rotation=45)
plt.grid(axis="y", linestyle="--", alpha=0.7)
plt.show()


```

```python
# Step 5.9 - Create DataFrame
data = {
    "first_name": ["Jason", "Molly", "Tina", "Jake", "Amy"],
    "last_name": ["Miller", "Jacobson", "Ali", "Milner", "Cooze"],
    "female": [0, 1, 1, 0, 1],
    "age": [42, 52, 36, 24, 73],
    "preTestScore": [4, 24, 31, 2, 3],
    "postTestScore": [25, 94, 57, 62, 70],
}

df = pd.DataFrame(data)

# Scatterplot with size determined by age
plt.figure(figsize=(6, 5), dpi=100)
plt.scatter(df["preTestScore"], df["postTestScore"], s=df["age"]*5, color="b", alpha=0.6)

plt.title("Scatter Plot of Pre-Test vs Post-Test Scores", fontsize=14)
plt.xlabel("Pre-Test Score", fontsize=14)
plt.ylabel("Post-Test Score", fontsize=14)
plt.grid(True, linestyle="--", alpha=0.7)
plt.show()
```

```python
#5.10
# Step 5.10 - Scatterplot with fixed size and color by gender
plt.figure(figsize=(6, 5), dpi=100)

# Assign colors: Red for female (1), Blue for male (0)
colors = ["red" if sex == 1 else "blue" for sex in df["female"]]

plt.scatter(df["preTestScore"], df["postTestScore"], s=300, c=colors, alpha=0.6)

plt.title("Scatter Plot with Color by Gender", fontsize=14)
plt.xlabel("Pre-Test Score", fontsize=14)
plt.ylabel("Post-Test Score", fontsize=14)
plt.grid(True, linestyle="--", alpha=0.7)

# Show the plot
plt.show()
```
