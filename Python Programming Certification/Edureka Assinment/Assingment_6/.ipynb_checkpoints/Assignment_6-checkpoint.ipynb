```python
#Assingement 3

import pandas as pd

# Load the CSV file into a Pandas DataFrame
df = pd.read_csv('SalaryGender.csv')

# Check the first few rows to understand the structure
print(df.head())

# Filter the data to keep only people with a Ph.D.
df_phd = df[df['PhD'] == 1]

# Select only the "Age" and "Ph.D." columns
df_phd_selected = df_phd[['Age', 'PhD']]

# Display the new DataFrame with Age and Ph.D. columns
print(df_phd_selected)

# Now, remove rows of people who don't have a Ph.D. from the original DataFrame
df = df[df['PhD'] == 1]

# Save the modified DataFrame back to a new CSV (optional, if you want to overwrite or save the updated file)
df.to_csv('filtered_salary_gender.csv', index=False)

# Show the updated DataFrame (optional)
print(df.head())
```

```python
#Assigment 7

import numpy as np

# Create a NumPy array with NaN values
arr = np.array([np.nan, 1., 2., np.nan, 3., 4., 5.])

# Print the original array
print("Original array:")
print(arr)

# Filter the array to omit NaN values
filtered_arr = arr[~np.isnan(arr)]

# Print the filtered array
print("\nArray with NaN values omitted:")
print(filtered_arr)
```

```python
#Assigment 10

import numpy as np

# Create a NumPy array with elements from 0 to 10
arr = np.arange(11)

# Negate elements between 3 and 9 (exclusive)
arr[(arr > 3) & (arr < 9)] *= -1

# Print the modified array
print("Modified Array:")
print(arr)
```

```python
#Assingment 11
import numpy as np

# Create a random 3x3 array
arr = np.random.rand(3, 3)

# Print the original array
print("Original Array:")
print(arr)

# Sort by the first column
sorted_by_first = arr[arr[:, 0].argsort()]

# Sort by the second column
sorted_by_second = arr[arr[:, 1].argsort()]

# Sort by the third column
sorted_by_third = arr[arr[:, 2].argsort()]

# Print the sorted arrays
print("\nSorted by First Column:")
print(sorted_by_first)

print("\nSorted by Second Column:")
print(sorted_by_second)

print("\nSorted by Third Column:")
print(sorted_by_third)
```

```python
#Assigment 12
import numpy as np

# Create a 4D array (e.g., shape (2, 3, 4, 5))
arr = np.random.rand(2, 3, 4, 5)

# Print the original 4D array
print("Original 4D Array:")
print(arr)

# Compute the sum over the last two axes (axes 2 and 3)
sum_over_last_two_axes = np.sum(arr, axis=(2, 3))

# Print the result of the summation
print("\nSum over the last two axes:")
print(sum_over_last_two_axes)
```

```python
#Assigmnet 14
import numpy as np

# Create a random 5x5 array
arr = np.random.rand(5, 5)

# Print the original array
print("Original Array:")
print(arr)

# Swap row 1 (index 0) with row 3 (index 2)
arr[[0, 2]] = arr[[2, 0]]

# Print the modified array
print("\nArray after swapping rows 1 and 3:")
print(arr)
```

```python
#Assignmet 15

import numpy as np

# Create a random 4x4 matrix
matrix = np.random.rand(4, 4)

# Print the original matrix
print("Original Matrix:")
print(matrix)

# Compute the rank of the matrix
rank = np.linalg.matrix_rank(matrix)

# Print the rank
print("\nRank of the Matrix:", rank)
```
