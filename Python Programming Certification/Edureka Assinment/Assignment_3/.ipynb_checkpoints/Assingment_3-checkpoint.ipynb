```python
# Assinment 1
class Animal:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def speak(self):
        return "The animal makes a sound."


class Dog(Animal):
    def __init__(self, name, age, breed):
    
        super().__init__(name, age)
        self.breed = breed

    def speak(self):
        return "Woof!"  # Override the speak method

my_dog = Dog("Buddy", 3, "Golden Retriever")

# Display the dog's details and the result of the speak method
print(f"My dog's name is {my_dog.name}, age is {my_dog.age}, and breed is {my_dog.breed}.")
print(f"My dog says: {my_dog.speak()}")
```

```python
# Assingment 2
class Vechile : 
    def __init__(self,name,color,wheel) :
        self.name=name
        self.color=color
        self.wheel=wheel

    def display_info(self):
        raise NotImplementedError

class Car(Vechile) :
    def __init__(self,name ,color,wheel=4):
        super().__init__(name,color,wheel)

    def display_info(self):
        return f"Car Name is {self.name} Color is {self.color} Number of wheel{self.wheel}"

class Bike(Vechile):

    def __init__(self ,name,color,wheel=2):
        super().__init__(name,color,wheel)

    def display_info(self):
        return f'Bike name is {self.name} color is {self.color} number of wheel is {self.wheel}'

bike=Bike('Hero Hoda','Red')
car = Car('Hoda City','White')

vehicles = [car, bike]
for vehicle in vehicles:
    print(vehicle.display_info())
```

```python
# Assingment 4
tank=0

def Fillup(fuel):
    global tank
    tank+=fuel
    return tank

def Use(fuel) :
    global tank
    if tank> fuel  :
        tank-=fuel
    else :
        print('Not Enough Fuel')
    return tank

x=Fillup(100)
print('Quantity of Fuel is ' +str(x) )

x=Fillup(100)
print('Quantity of Fuel is ' +str(x) )

y=Use(10)
print('quantity of fuel is ' + str(y))
```

```python
#Assingement 5

# Base class 1
class Class1:
    def __init__(self):
        print("Class 1 Constructor")
    
    def display_class1(self):
        print("This is Class 1")

# Class 2 inherits from Class 1
class Class2(Class1):
    def __init__(self):
        super().__init__()
        print("Class 2 Constructor")
    
    def display_class2(self):
        print("This is Class 2")

# Class 3 inherits from Class 1
class Class3(Class1):
    def __init__(self):
        super().__init__()
        print("Class 3 Constructor")
    
    def display_class3(self):
        print("This is Class 3")

# Class 4 inherits from Class 2 and Class 3
class Class4(Class2, Class3):
    def __init__(self):
        # Call constructors of both Class2 and Class3
        Class2.__init__(self)
        Class3.__init__(self)
        print("Class 4 Constructor")
    
    def display_class4(self):
        print("This is Class 4")

# Create an object of Class 4
obj = Class4()

# Call methods of all classes through Class 4 object
obj.display_class1()  # Inherited from Class 1
obj.display_class2()  # Inherited from Class 2
obj.display_class3()  # Inherited from Class 3
obj.display_class4()  # Defined in Class 4
              
        
```

```python
# Assingment 6

class Person:
    def __init__(self):
        print('Inside Constructor')
        
    def set_name(self , name):
        self.name=name
    def set_age(self,age):
        self.age=age

    def get_name(self):
        return self.name
    def get_age(self):
        return self.age

person=Person()
person.set_name('Abhay')
person.set_age(22)
print(f' Name is {person.get_name()} ')
print(f'Age is {person.get_age()}')
```
