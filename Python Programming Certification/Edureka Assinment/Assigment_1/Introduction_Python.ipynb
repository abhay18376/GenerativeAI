```python
def Factor_of_Nubmer(number):
    
    for i in range(1, number + 1):
        # Check if i is a factor of the number
        if number % i == 0:
            # Check if the factor is even or odd
            if i % 2 == 0:
                print(f"{i} is an even factor of {number}")
            else:
                print(f"{i} is an odd factor of {number}")

# Input from the user
num = int(input("Enter a number: "))
Factor_of_Nubmer(num)
```

```python
def sort_words():
    
    input_string = input("Enter a sequence of words: ")
    words_list = input_string.split()

    words_list.sort()
    print("Sorted words:", " ".join(words_list))


sort_words()
```

```python
def find_odd_digit_numbers():
    count = 0  # To count numbers with all odd digits
    
    # Iterate through numbers from 1001 to 2999 (both 1000 and 3000 are excluded)
    for num in range(1001, 3000):
        # Convert the number to a string to check each digit
        str_num = str(num)
        
        # Check if each digit is odd
        if all(int(digit) % 2 != 0 for digit in str_num):
            count += 1  # Increment count if all digits are odd
    
    # Print the count of such numbers
    print("Count of numbers with all odd digits:", count)

# Call the function
find_odd_digit_numbers()
        
```

```python
def LetterCount(str) :
    countchar=0
    countnum=0

    for c in  str:
        if c.isalpha(): 
            countchar+=1
        if c.isdigit() :
            countnum+=1
    print(countchar)         
    print(countnum)
str=input('Enter the string')
LetterCount(str)
```

```python
def isPallidium(num):
    revstr=num[::-1]
    if revstr == num :
        print(f'Is a Palladuim number {num}')
    else :
        print(f'Not a Palliduim number {num}')
num=input('Enter the number')
isPallidium(num)
```
