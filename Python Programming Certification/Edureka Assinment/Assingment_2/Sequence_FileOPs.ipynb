```python
import re
def Verify_the_Password(password):
    if len(password)<5 or len(password)>10 :
        return "Password lenght should  be between 5 and 10"
    if not re.search("[a-zA-Z]", password):
        return "Atleast one alphabet Required"
    if not re.search("[0-9]",password):
        return "Atleast one number is Required"
    if not re.search("[@&]",password):
        return "Atleast one character is Required"
    return "Valid Password"
password=input('Enter the Password')
res=Verify_the_Password(password)
print(res)
```

```python
def get_even_indexed_alphabets(input_string):
    result = ""
    for index in range(0, len(input_string), 2):  
        if input_string[index].isalpha():  
            result += input_string[index]
    return result

listinput=input('enter th input')
result = get_even_indexed_alphabets(listinput)
print(result)
```

```python
def print_char_count(input_string):
    char_count = {}
    
    for char in input_string:
        if char in char_count:
            char_count[char] += 1
        else:
            char_count[char] = 1
    
    for char, count in char_count.items():
        print(f"Character: {char}, Count: {count}")

user_input = input("Enter a string: ")
print_char_count(user_input)
```

```python
import hashlib

def is_valid_reference_id(reference_id):
    if len(reference_id) == 12 and reference_id.is():
        return True
    return False


def encrypt_reference_id(reference_id):
    sha256_hash = hashlib.sha256()
    sha256_hash.update(reference_id.encode('utf-8'))
    return sha256_hash.hexdigest()

def main():
    reference_id = input("Enter your Reference ID (12 characters, alphanumeric): ")
  
    if is_valid_reference_id(reference_id):
        print("Reference ID is valid.")
        
        encrypted_reference_id = encrypt_reference_id(reference_id)
        
        print(f"Encrypted Reference ID: {encrypted_reference_id}")
    else:
        print("Invalid Reference ID. Please ensure it is 12 characters long and contains only alphanumeric characters.")

# Run the program

main()
```
