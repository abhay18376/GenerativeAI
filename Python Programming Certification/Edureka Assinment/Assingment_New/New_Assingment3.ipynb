```python
# 2 
class Car: 
    def __init__(self,name,color,wheel):
        self.name=name
        self.color=color
        self.wheel=wheel

    def Display(self): 
        print(f'Car Name {self.name} Colour of Car {self.color} No of Wheel {self.wheel}')

    def Manufacture_Name(self):
        print('Manufacture Name Toyota')

class Bike(Car):
    def __init__(self,name,color,wheel=2):
        super().__init__(name,color,wheel)
        
    def Manufacture_Name(self):
        print('Manufacture Name Hero')

bike=Bike('CVR','Blue')
bike.Display()
bike.Manufacture_Name()
```

```python
# 4
Number_of_Liter=10

def Fill_Petrol(qnty): 
    global Number_of_Liter
    Number_of_Liter= Number_of_Liter + qnty
    return Number_of_Liter
def Use_Petrol(qnty): 
    global Number_of_Liter 
    Number_of_Liter= Number_of_Liter - qnty
    return Number_of_Liter

liter=Use_Petrol(1)
print(f'Number of Petrol remaining {liter}')
```

```python
# 5
class Class1:
    def print_message(self):
        print("This is Class1")

# Class2 inherits from Class1
class Class2(Class1):
    def print_message(self):
        print("This is Class2")

# Class3 inherits from Class1
class Class3(Class1):
    def print_message(self):
        print("This is Class3")

# Class4 inherits from Class2 and Class3 (Multiple Inheritance)
class Class4(Class2, Class3):
    def print_message(self):
        print("This is Class4")
```

```python
# Creating objects
obj1 = Class1()
obj2 = Class2()
obj3 = Class3()
obj4 = Class4()

# Calling print_message from each object
obj1.print_message()  # Output: This is Class1
obj2.print_message()  # Output: This is Class2
obj3.print_message()  # Output: This is Class3
obj4.print_message()  # Output: This is Class4
```

```python
# 6 
class Person: 
    def __init__(self): 
        self._name=None
        self._age=None

    def set_name(self,name): 
        self._name=name
    def set_age(self,age):
        self._age=age

    def get_name(self): 
        return self._name
    def get_age(self): 
        return self._age

obj1=Person()
obj1.set_name("Mathew")
obj1.set_age(35)

print(f'Name of Person is {obj1.get_name()}')
print(f'Age of Person is {obj1.get_age()}')
```
