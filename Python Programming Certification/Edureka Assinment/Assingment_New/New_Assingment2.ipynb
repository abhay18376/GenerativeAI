```python
# 1 

def ValidPassword(password): 
    has_alpha = any(char.isalpha() for char in password)
    
    has_integer=any(c.isnumeric() for c in password)
    
    has_atleast=any(c in ['@','&']for c in password) 
    
    if((len(password)>5) & (len(password)<=10)): 
        has_lenght=True
    else :
        has_lenght=False
    
    if((has_alpha==True) & (has_integer==True) & (has_atleast==True) &(has_lenght==True)):
        print('Password is Valid')
    else:
        print('Password is Invalid')
```

```python
ValidPassword('Abhay1234&')
```

```python
# 11
from cryptography.fernet import Fernet
num=input('Enter ten digit number')
if(num.isdigit()):
    # Generate a key
    key = Fernet.generate_key()
    cipher_suite = Fernet(key)

    # Encrypt a message
    message = num.encode()
    cipher_text = cipher_suite.encrypt(message)
    print(cipher_text)
else :
    print('Invalid number')
```
