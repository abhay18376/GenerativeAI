```python
# No 1
def Fact_number(n): 
    fact=1
    for i in range(1,n+1): 
        fact=fact*i
    print(f'Factorial of {n} is {fact}')
    if(fact % 2 ==0):
        print('Factorial is Even')
    else :
        print('Factorial is Odd')
        
    
```

```python
Fact_number(9)
```

```python
# No 2
def Sort_word(str2): 
    spllst=str2.split()
    spllst.sort()
    return spllst
```

```python
# No 3 
def Odd_Num(num): 
    digits=[int(d) for d in str(num)]
    iseven = list(filter(lambda x : x % 2!=0,digits))
    if(len(iseven)==4):
        print(num)
for n in range(1000,3000): 
    Odd_Num(n)
```
