```python
#1
import pandas as pd

# Load the CSV file
file_path = "sample-salesv2.csv"
df = pd.read_csv(file_path)

# Display basic information about the dataset
df.head()


```

```python
# Summary statistics for 'unit price' column
unit_price_summary = df["unit price"].describe()
unit_price_summary

```

```python
# filter the data
filtered_df = df[["name", "net_price", "date"]]

# Group by 'name' and aggregate net_price
grouped_df = filtered_df.groupby("name").agg({"net_price": "sum", "date": "count"}).reset_index()

# Rename columns for clarity
grouped_df.rename(columns={"net_price": "total_net_price", "date": "transaction_count"}, inplace=True)

# Display the first few rows of the grouped DataFrame
grouped_df.head()
```

```python
#Plot Graph
import matplotlib.pyplot as plt

# Sort by total sales for better visualization
grouped_df_sorted = grouped_df.sort_values(by="total_net_price", ascending=False)

# Plot the graph
plt.figure(figsize=(15, 6))
plt.bar(grouped_df_sorted["name"], grouped_df_sorted["total_net_price"], color="skyblue")

# Customize labels and title
plt.xlabel("Customer Name", fontsize=12)
plt.ylabel("Total Sales ($)", fontsize=12)
plt.title("Total Sales by Customer", fontsize=14)
plt.xticks(rotation=90)  # Rotate x-axis labels for readability

# Show the plot
plt.show()
```

```python
#Create a DataFramefromthefollowing data:

import pandas as pd

# Define data dictionary
data = {
    'first_name': ['Jason', 'Molly', 'Tina', 'Jake', 'Amy'],
    'last_name': ['Miller', 'Jacobson', 'Ali', 'Milner', 'Cooze'],
    'female': [0, 1, 1, 0, 1],
    'age': [42, 52, 36, 24, 73],
    'preTestScore': [4, 24, 31, 2, 3],
    'postTestScore': [25, 94, 57, 62, 70]
}

# Create DataFrame
df_people = pd.DataFrame(data)

# Display the DataFrame
df_people
```

```python
#5.10: Draw a Scatterplot from the data in question 9 of preTestScore and postTestScore with the size = 300 and the color determined by sex


# Define color mapping based on 'female' column (0 = blue, 1 = red)
colors = ['blue' if gender == 0 else 'red' for gender in df_people['female']]

# Create scatter plot
plt.figure(figsize=(6, 5), dpi=100)
plt.scatter(df_people['preTestScore'], df_people['postTestScore'], s=300, c=colors, alpha=0.6, edgecolors='black')

# Add title and labels
plt.title("Scatter Plot of Pre-Test vs Post-Test Scores", fontsize=14)
plt.xlabel("Pre-Test Score", fontsize=14)
plt.ylabel("Post-Test Score", fontsize=14)

# Show legend
plt.scatter([], [], c='blue', s=300, label="Male")  # Dummy scatter points for legend
plt.scatter([], [], c='red', s=300, label="Female")
plt.legend()

# Show plot
plt.grid(True, linestyle="--", alpha=0.6)
plt.show()
```
