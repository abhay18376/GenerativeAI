```python
# Universal Function 
1. Mathematical 
2. Trigonometry
3. bit dwiddling or bit manipulation
4. Floating function
5. Comparision
```

```python
# MAthematical Function 
nar1=np.array([[4,5],[7,6]])
nar2=np.array([[2,3],[5,6]])
print(nar1)
print(nar2)
print(nar1+nar2)
np.add(nar1 ,nar2)
```

```python
print(np.subtract(nar1,nar2))
```

```python
np.lcm(nar1,nar2)
```

```python
np.sqrt(nar1)
```

```python
#Triginometry

tarr1=np.array([[2,1],[5,4]])
tarr1

print(np.sin(tarr1))
print(np.cos(tarr1))
```

```python
#bit dwiddling or bit manipulation
barr1=np.array([[1,2],[2,1]])
barr2=np.array([[3,4],[4,3]])

print(np.bitwise_and(barr1,barr2))
print(np.bitwise_or(barr1,barr2))
```

```python
#Comparision 
carr1=np.array([[3,1],[1,1]])
carr2=np.array([[4,6],[5,7]])
print(carr1<carr2)
a= np.arange(4).reshape(2,2)
a
np.greater(carr1,carr2)
```

```python
ar1=np.array([1.2,2.1,3.2,np.log(-1)])
ar1
np.isnan(ar1)
np.floor(ar1)
```

```python
# Aggreation
arr=np.arange(1,10)
print(arr)
#Aggregate function
print(np.sum(arr))
print(np.mean(arr))
print(np.min(arr))
print(np.max(arr))
a22=[[3,4],[1,0]]
np.all(a22)  # all the value are non zero
np.any(a22)
```

```python
# Splitting array 
arrr=np.array([1,2,3,4,5,6,7,8])
print(arrr)
print(np.split(arrr,4))
are=np.reshape(arrr,(4,2))
np.hsplit(are,2)
```

```python
# opening and writing in binary mode 
fp=open('file1_bin.txt','ab+')
arr=np.array([[1,2],[3,4]])
print(arr)
np.save(fp,arr)
```

```python
fp.seek(0)
arr=np.load(fp)
print(arr)
```
