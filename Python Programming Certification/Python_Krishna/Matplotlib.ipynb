```python
import matplotlib.pyplot as plt 
import numpy as np
import pandas as pd 
rainfall=pd.read_csv('Class Note/dataset/rainfall_in_kerala.csv',index_col=0)
rainfall
```

```python
rainfall.columns
```

```python
print(rainfall['ANNUAL'].loc['1995':'2000'])
```

```python
plt.plot(rainfall['ANNUAL'].loc['1995':'2000'])
plt.title('Kerala Rain')
plt.xlabel('Year')
plt.ylabel('Rainfall Yearly')
plt.show()
```

```python
plt.figure(figsize=(9,9))
plt.plot(rainfall.loc[1995,'JAN':'DEC'],"b",label='1996 Rainfall')
plt.plot(rainfall.loc[2000,'JAN':'DEC'],"c",label='2000 Rainfall')
plt.xlabel('Month')
plt.ylabel('Rainfall')
plt.title('Rainfall from 1995 2000')
plt.legend(loc='upper left')
plt.savefig('Rainfall_Graph.jpeg')
plt.show()
```

```python
for i in rainfall.columns: 
    if(i!="ANNUAL"): 
        plt.barh(i,rainfall[i].sum())
plt.savefig('Rainfall_bar.jpeg')
plt.plot()
    
```

```python
plt.hist(rainfall.APR)
plt.xlabel("April Month Values")
plt.ylabel("Frequencie")
plt.show()
```

```python
# plt.pie (aggregate value, label)
plt.figure(figsize=(9,9))
month_wise_sum=list()
month_name=[]
for i in rainfall.columns: 
    if(i!='ANNUAL'): 
        month_name.append(i)
        month_wise_sum.append(rainfall[i].sum())
plt.pie(month_wise_sum,labels=month_name,rotatelabels=True,autopct="%.2f")
plt.show()
        
    
```

```python
plt.scatter(rainfall['JAN'],rainfall['FEB'])
plt.show()
```

```python
plt.boxplot(rainfall['JAN'])
plt.show()
```

```python
cereal=pd.read_csv('Class Note/dataset/cereal.csv')
cereal
```

```python
for col in cereal.columns[3:7]: 
    print(col)
```

```python
index=1
for col in cereal.columns[3:7]: 
    plt.subplot(2,2, index)  # index is used to generate unique number
    plt.plot(cereal[col])
    plt.title(col)
    index=index+1
plt.show()
```

```python
my_grid=plt.GridSpec(3,3)
myfig=plt.figure(figsize=(5,5))

myfig.add_subplot(my_grid[0:2,0:2])
myfig.Title='G1'
plt.plot(cereal.sodium)

myfig.add_subplot(my_grid[0:2,2])
myfig.Title='G2'
plt.barh(cereal.index,cereal.rating)

myfig.add_subplot(my_grid[2,0:3])
myfig.Title='G3'
plt.plot(cereal.fat)

plt.show()
```

```python
y = [20, 21, 20.5, 20.8]
X = [1,2,3,4] 

plt.figure(figsize=(8, 4))
plt.plot(X, y, marker='o', color='blue', linestyle='-', linewidth=2, markersize=8)

plt.title("Simple Line Plot")
plt.xlabel("X values")
plt.ylabel("y values")
plt.grid(True, alpha=0.3)
plt.show()
```
