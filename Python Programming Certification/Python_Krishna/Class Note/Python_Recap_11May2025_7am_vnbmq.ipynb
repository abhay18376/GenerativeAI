```python
class Student:
  # student name, rollno - data

  # self.name, self.no - data
  # set_name, set_no, get_name, get_no - functions

  # class variable
  no_of_students = 0

  def __init__(self):
    print('constructor called...')
    Student.no_of_students = Student.no_of_students + 1

  # setters
  def set_name(self, a):
    self.name = a
  def set_no(self, b):
    self.no = b

  # getters
  def get_name(self):
    return self.name
  def get_no(self):
    return self.no
```

```python
def main():
  print('enter first number? ')
  n1 = eval(input())
  print('enter second number? ')
  n2 = eval(input())

  try: # guessing this code might throw error
    result = n1 / n2
    # some other code
    x = []
    #x[9] = 1000
    a = "krishna" + 1
  except ZeroDivisionError: # when the code failed
    print('looks like you are passing second number as zero which is not allowed. Please check')
  except IndexError:
    print('Looks like you are tyring to add element to list without append or extend or insert function')
  except Exception as obj: # generic exception
    print('some error hapened which we are not aware...', str(obj))
  else: # code successfully execute
    print('Division result is : ', result)
  # after 200 lines of code
  print('Program ended...')
main()

```
