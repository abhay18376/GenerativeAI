```python
def Numberdemo(num):
    if(num>0):
        print('Number is Positive')
    elif(num<0):
        print('Number is Negative')
    else:
        print('Number is zero')
    
number = eval(input('Enter the Number'))
Numberdemo(number)
```

```python
def sumdemo(a,b,c,d):
    sum=a+b+c+d
    return sum,100

result=sumdemo(11,22,33,44)
print(result)
```

```python
# Default Argument Function
def MultipleDemo(a,b=10):
    result = a * b
    return result

MultipleDemo(11,33)
MultipleDemo(2)
```

```python
#Keyword Argument
def keywordfunc(a,b):
    result= a+b
    return result

keywordfunc(b=3, a=222)
```

```python
# variable Argument
def va(*value):
    x=1
    for i in value:     
        x=x * i
        
    return x

va(11,22,33,44)
va(1,2,3)
```

```python
# Local and Global variable (Scope of varaible)
X=100
def Sumdemo():
    global X
    X=200
    print(X)
    
Sumdemo()
print(X)
```
