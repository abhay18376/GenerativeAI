# lambda is use for short term solution

```python
sqrt=lambda x: x*2 
sqrt(22)

sumdemo=lambda a,b,c : a+b+c
sumdemo(11,22,33)
```

```python
lst=[1,2,3,4,6]

sqrt=list(map(lambda x : x*3,lst))
sqrt

list(map(lambda z:z**2,lst))
```

```python
fltr=list(filter(lambda x: x>2,lst))
fltr
list(filter(lambda x:x >0,lst))
```
