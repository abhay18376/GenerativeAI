```python
# creating ddataframe 
import pandas as pd 
data=pd.DataFrame({"Product":['Carrot','Bannana','Beans','Orange','Barcoali','Bannana'], 
                 "Category":['Vegetable','Fruit','Vegetable','Fruit','Vegetable','Fruit'], 
                 "Quantity":[20,30,40,50,60,80], 
                 "Amount" : [100,200,400,500,200,300]})

pivot=data.pivot_table(index='Product',values='Amount',aggfunc='sum')
pivot
```
