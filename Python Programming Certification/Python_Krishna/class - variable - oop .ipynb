```python
class Student:
    total_student_join=0                    # class variable
    #data variable name,grade ( also called instance variable, object variable  )

    # Constructor
    def __init__(self):
        Student.total_student_join =Student.total_student_join + 1

    def __del__(self):
        print('Destructor Called')
    
    def set_student(self,name,grade):       #Setter
        self.name=name                      # Current Object:self
        self.grade=grade
        
    
    def get_student(self):
        return self.name,self.grade         #getter

s1=Student()
s1.set_student('Abhay','Distinction')

s2=Student()
s2.set_student('Ashok','Distinction')
print(f'Total Student Join {Student.total_student_join}')
#del(s1)

print (Student.__doc__)
```

# Access Control controlling the acess of data and function
1. Public
2. Private
3. Protected

```python
class Access:
    def __init__(self):
        self.num=10
        self.__num2 =100
        self._num3 =200
    def disp(self):
        return self.__num2
    def __Privatefunction(self):
        print('Private Function')  
    def _ProtectedFunction(self):
        print('Protected Function')
        
acc=Access()
print(acc.num)
acc.disp()
#  print(acc.num3) protected variable can access in child class 
#acc.____Privatefunction()
acc._ProtectedFunction()
```

```python
# Multiple Inheritance
class Ecommerce:
    def __init__(self):
        print("Ecommerce") 
    def get_board_of_director(self):
        print('Jeff, Steve')
    def get_company_headquater(self):
        print('New York , New Jersey')

class Food_Division_Management:
    def __init__(self):
        print("Food_Division_Management")
    def get_ceo_name(self):
        print('George, Mark')
    def get_company_headquater(self):
        print('Mumbai' , 'Chenai')
         
class FoodDivision(Ecommerce,Food_Division_Management):
    def __init__(self):
        #return super().__init__()
        return Food_Division_Management.__init__(self)
        print("FoodDivision")
    def get_company_headquater(self):
        print('Singapore')
    
fd=FoodDivision()
fd.get_board_of_director()
fd.get_company_headquater()
fd.get_ceo_name()
```
