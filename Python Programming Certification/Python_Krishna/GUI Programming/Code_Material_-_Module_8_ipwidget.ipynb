```python
# Getting the rating score - event driven way

a_intslider = widgets.IntSlider(
value=6, # default value
min=0,
max=20,
step=3,
description='a',
disabled=False,
continuous_update=True,
orientation='vertical',
readout=True,
readout_format="d") # d - int, f = float

def f(a):
    print('Given Rating is:', a)

out = widgets.interactive_output(f, {'a':a_intslider})

widgets.VBox([widgets.VBox([a_intslider]), out])
```

```python
# A simple Calculator program

from ipywidgets import Button, HBox, VBox, Text
from functools import reduce

words = ['1', '2', '3', '4', '5', '6', '7', '8', '9', '+', '*', '-'] 

items = [Button(description=w) for w in words] 
print(items)

first_row_box = HBox([items[i] for i in range(3)])
second_row_box = HBox([items[i] for i in range(3,6)])
third_row_box = HBox([items[i] for i in range(6, 9)])
fourth_row_box = HBox([items[i] for i in range(9,len(items))])
#print(first_row_box)
text_box = Text(description = 'Result')

layout1 = VBox([first_row_box, third_row_box, second_row_box, fourth_row_box])
layout2 = VBox([text_box])
display(layout1, layout2)


nums = []
def on_button_clicked(b):
    global nums
    
    print(nums)
    if b.description not in ('+', '*', '-'):
        nums.append(int(b.description))
    else:
        if b.description == '+':
            result = sum(nums)
            text_box.value = str(result)
        elif b.description == '*':
            result = reduce((lambda a, b: a * b), nums)
            text_box.value = str(result)
        elif b.description == '-':
            result = reduce((lambda a, b: a - b), nums)
            text_box.value = str(result)
        nums.clear()

for i in range(len(items)):
    items[i].on_click(on_button_clicked)
    
```
