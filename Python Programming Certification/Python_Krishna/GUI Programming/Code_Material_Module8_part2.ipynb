## Button

```python
# Example 18
# Button

from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
from ipywidgets import widgets
from IPython.display import display

button = widgets.Button(description='Click me',
disabled=False,
button_style='success',tooltip='You have to click here...',
icon='check')

display(button)
def on_button_clicked(b):
    driver = webdriver.Chrome('/Users/bonkv/Class/chromedriver')
    driver.maximize_window()
    driver.implicitly_wait()
    
    driver.get('http://www.python.org')
    search_textbox = driver.find_element_by_class_name('search-field')
    search_textbox.clear()
    search_textbox.send_keys('python')
    
    go_button = driver.find_element_by_id('submit')
    go_button.click()
    
    search_results = driver.find_elements_by_xpath('//ul/li/h3')
    for each_result in search_results:
        print(each_result.text)
    driver.quit()
    #time.sleep(100)
  
    
button.on_click(on_button_clicked)
```

```python
# program to identify given binary numbers divisible by 5 or not. 
x = ['0100' ,'1000' ,'1010']

for bin_num in x:
    dec_num = int(bin_num, 2)
    if dec_num % 5 == 0:
        print('binary number %s is divisible by 5' % str(bin_num))
```
