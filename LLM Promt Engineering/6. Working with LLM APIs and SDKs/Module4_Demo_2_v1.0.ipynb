## Step 4: Understanding Generation Configuration
Display and explain default generation parameters and their meanings.
- **Temperature**: Controls randomness (0.0 = deterministic, 1.0 = highly creative)
- **Top-p**: Nucleus sampling threshold (focuses on most probable tokens)
- **Top-k**: Limits vocabulary to top K most likely tokens
- **Max tokens**: Maximum length of generated response
- **Stop sequences**: Specific text patterns that end generation

```python
def display_default_config():
    """Display default generation configuration"""
    print("📋 Default Gemini Generation Configuration")
    print("=" * 45)

    # Create default config
    default_config = genai.types.GenerationConfig()

    config_dict = {
        "temperature": getattr(default_config, 'temperature', 'Not set'),
        "top_p": getattr(default_config, 'top_p', 'Not set'),
        "top_k": getattr(default_config, 'top_k', 'Not set'),
        "max_output_tokens": getattr(default_config, 'max_output_tokens', 'Not set'),
        "stop_sequences": getattr(default_config, 'stop_sequences', 'Not set')
    }

    for param, value in config_dict.items():
        print(f"   {param}: {value}")

    print("\n📚 Parameter Explanations:")
    print("   • Temperature: Controls randomness (0.0 = deterministic, 1.0 = creative)")
    print("   • Top-p: Nucleus sampling threshold (0.1 = focused, 0.9 = diverse)")
    print("   • Top-k: Limits vocabulary to top K tokens")
    print("   • Max tokens: Maximum length of generated response")
    print("   • Stop sequences: Tokens that end generation")

display_default_config()
```

## Step 5: Temperature Parameter Deep Dive
Compare different temperature values (0.0, 0.3, 0.7, 1.0) using the same creative writing prompt to show how randomness affects output creativity and consistency.

```python
def temperature_comparison():
    """Compare different temperature settings with the same prompt"""
    print("🌡️  Temperature Parameter Comparison")
    print("=" * 45)

    model = genai.GenerativeModel('gemini-2.0-flash')
    base_prompt = "Write a creative story about a robot learning to paint. Make it exactly 3 paragraphs."

    temperatures = [0.0, 0.3, 0.7, 1.0]

    print(f"🎯 Base Prompt: {base_prompt}")
    print("\n" + "="*80)

    for temp in temperatures:
        print(f"\n🌡️  TEMPERATURE: {temp}")
        print("-" * 40)

        config = genai.types.GenerationConfig(
            temperature=temp,
            max_output_tokens=300
        )

        try:
            response = model.generate_content(
                base_prompt,
                generation_config=config
            )

            print(f"📝 Generated Response:")
            print(response.text)
            print(f"\n📊 Response Length: {len(response.text)} characters")
            print(f"📊 Word Count: {len(response.text.split())} words")

        except Exception as e:
            print(f"❌ Error: {e}")

        print("\n" + "="*80)
        time.sleep(1)

temperature_comparison()

```

## Step 6: Top-P (Nucleus Sampling) Demonstration
Test various top-p values (0.1, 0.5, 0.8, 0.95) to demonstrate how nucleus sampling affects response diversity and focus.
- **Low top-p (0.1)**: Very focused, conservative responses
- **High top-p (0.95)**: More diverse, varied vocabulary


```python
def top_p_comparison():
    """Compare different top-p values"""
    print("🎯 Top-P (Nucleus Sampling) Comparison")
    print("=" * 45)

    model = genai.GenerativeModel('gemini-2.0-flash')
    base_prompt = "Explain quantum computing in simple terms. Include 3 key concepts and their practical applications."

    top_p_values = [0.1, 0.5, 0.8, 0.95]

    print(f"🎯 Base Prompt: {base_prompt}")
    print("\n" + "="*80)

    for top_p in top_p_values:
        print(f"\n🎯 TOP-P: {top_p}")
        print("-" * 40)

        config = genai.types.GenerationConfig(
            top_p=top_p,
            temperature=0.7,  # Keep temperature constant
            max_output_tokens=400
        )

        try:
            response = model.generate_content(
                base_prompt,
                generation_config=config
            )

            print(f"📝 Generated Response:")
            print(response.text)
            print(f"\n📊 Response Length: {len(response.text)} characters")
            print(f"📊 Unique Words: {len(set(response.text.lower().split()))} words")

        except Exception as e:
            print(f"❌ Error: {e}")

        print("\n" + "="*80)
        time.sleep(1)

top_p_comparison()
```

## Step 8: Max Output Tokens Control
Demonstrate how max_output_tokens parameter (50, 150, 300, 500) controls response length and content completeness.


```python
def max_tokens_comparison():
    """Demonstrate max_output_tokens parameter"""
    print("📏 Max Output Tokens Comparison")
    print("=" * 40)

    model = genai.GenerativeModel('gemini-2.0-flash')
    base_prompt = "Explain the history of the internet, from its origins to modern day. Include major milestones and key technologies."

    token_limits = [50, 150, 300, 500]

    print(f"🎯 Base Prompt: {base_prompt}")
    print("\n" + "="*80)

    for max_tokens in token_limits:
        print(f"\n📏 MAX OUTPUT TOKENS: {max_tokens}")
        print("-" * 40)

        config = genai.types.GenerationConfig(
            max_output_tokens=max_tokens,
            temperature=0.5
        )

        try:
            response = model.generate_content(
                base_prompt,
                generation_config=config
            )

            print(f"📝 Generated Response:")
            print(response.text)
            print(f"\n📊 Actual Length: {len(response.text)} characters")
            print(f"📊 Word Count: {len(response.text.split())} words")
            print(f"📊 Estimated Tokens: ~{len(response.text.split()) * 1.3:.0f}")

        except Exception as e:
            print(f"❌ Error: {e}")

        print("\n" + "="*80)
        time.sleep(1)

max_tokens_comparison()
```

## Step 9: Stop Sequences Demonstration
Show how stop sequences can control when generation ends, useful for structured outputs and preventing unwanted content continuation.


```python
def stop_sequences_demo():
    """Demonstrate stop sequences functionality"""
    print("🛑 Stop Sequences Demonstration")
    print("=" * 40)

    model = genai.GenerativeModel('gemini-2.0-flash')

    # Different stop sequence scenarios
    scenarios = [
        {
            "name": "No Stop Sequences",
            "prompt": "List the planets in our solar system with descriptions:",
            "stop_sequences": None
        },
        {
            "name": "Stop at numbered list item",
            "prompt": "List the planets in our solar system with descriptions:",
            "stop_sequences": ["5."]
        },
        {
            "name": "Stop at specific word",
            "prompt": "Write a story about space exploration. The story should mention Mars.",
            "stop_sequences": ["Mars"]
        },
        {
            "name": "Multiple stop sequences",
            "prompt": "Explain programming concepts: variables, functions, loops, and classes.",
            "stop_sequences": ["loops", "classes"]
        }
    ]

    for scenario in scenarios:
        print(f"\n🛑 SCENARIO: {scenario['name']}")
        print(f"🎯 Prompt: {scenario['prompt']}")
        print(f"🛑 Stop Sequences: {scenario['stop_sequences']}")
        print("-" * 60)

        config = genai.types.GenerationConfig(
            temperature=0.7,
            max_output_tokens=400,
            stop_sequences=scenario['stop_sequences']
        )

        try:
            response = model.generate_content(
                scenario['prompt'],
                generation_config=config
            )

            print(f"📝 Generated Response:")
            print(response.text)
            print(f"\n📊 Response Length: {len(response.text)} characters")

        except Exception as e:
            print(f"❌ Error: {e}")

        print("\n" + "="*80)
        time.sleep(1)

stop_sequences_demo()
```

## Step 10: Combined Parameters Optimization
Demonstrate three parameter combinations for different use cases:
- **Conservative**: Low temperature/top-p for formal, predictable text
- **Balanced**: Medium settings for professional flexibility
- **Creative**: High settings for expressive, varied outputs


```python
def combined_parameters_demo():
    """Show how parameters work together"""
    print("🎛️  Combined Parameters Optimization")
    print("=" * 45)

    model = genai.GenerativeModel('gemini-2.0-flash')
    base_prompt = "Create a professional email responding to a job interview invitation. Be polite and professional."

    configurations = [
        {
            "name": "Conservative (Formal/Predictable)",
            "config": {
                "temperature": 0.2,
                "top_p": 0.4,
                "top_k": 20,
                "max_output_tokens": 200
            }
        },
        {
            "name": "Balanced (Professional/Flexible)",
            "config": {
                "temperature": 0.5,
                "top_p": 0.7,
                "top_k": 40,
                "max_output_tokens": 250
            }
        },
        {
            "name": "Creative (Expressive/Varied)",
            "config": {
                "temperature": 0.8,
                "top_p": 0.9,
                "top_k": 80,
                "max_output_tokens": 300
            }
        }
    ]

    print(f"🎯 Base Prompt: {base_prompt}")
    print("\n" + "="*80)

    for config_set in configurations:
        print(f"\n🎛️  CONFIGURATION: {config_set['name']}")
        print(f"⚙️  Settings: {json.dumps(config_set['config'], indent=2)}")
        print("-" * 60)

        config = genai.types.GenerationConfig(**config_set['config'])

        try:
            response = model.generate_content(
                base_prompt,
                generation_config=config
            )

            print(f"📝 Generated Response:")
            print(response.text)
            print(f"\n📊 Response Length: {len(response.text)} characters")
            print(f"📊 Word Count: {len(response.text.split())} words")

        except Exception as e:
            print(f"❌ Error: {e}")

        print("\n" + "="*80)
        time.sleep(1)

combined_parameters_demo()
```

## Step 11: Parameter Impact Analysis Tool
Build a reusable tool to systematically test parameter effects and analyze their impact on response characteristics.


```python
class ParameterAnalyzer:
    """Tool to analyze parameter impact on generation"""

    def __init__(self):
        self.model = genai.GenerativeModel('gemini-2.0-flash')
        self.results = []

    def test_parameter(self, prompt: str, param_name: str, values: list):
        """Test a specific parameter with different values"""
        print(f"🔬 Testing {param_name.upper()} Parameter")
        print("=" * 40)

        for value in values:
            config_dict = {param_name: value}
            config = genai.types.GenerationConfig(**config_dict)

            try:
                response = self.model.generate_content(prompt, generation_config=config)

                result = {
                    'parameter': param_name,
                    'value': value,
                    'response_length': len(response.text),
                    'word_count': len(response.text.split()),
                    'response': response.text[:100] + "..." if len(response.text) > 100 else response.text
                }

                self.results.append(result)
                print(f"{param_name}={value}: {result['word_count']} words, '{result['response']}'")

            except Exception as e:
                print(f"{param_name}={value}: Error - {e}")

            time.sleep(0.5)

    def get_summary(self):
        """Get analysis summary"""
        if not self.results:
            return "No results to analyze"

        summary = "📊 Parameter Analysis Summary:\n"
        for result in self.results[-3:]:  # Show last 3 results
            summary += f"   {result['parameter']}={result['value']}: {result['word_count']} words\n"

        return summary

# Test the analyzer
analyzer = ParameterAnalyzer()
test_prompt = "Explain machine learning in one paragraph."

analyzer.test_parameter(test_prompt, "temperature", [0.1, 0.5, 0.9])
print(analyzer.get_summary())

```
