## Step 4: Explore Available Models
Create a function to discover and display all available Gemini models with their specifications and capabilities.

```python
def explore_gemini_models():
    """Explore and display available Gemini models"""
    print("🔍 Exploring Gemini Models:")
    print("=" * 50)

    for model in genai.list_models():
        if 'generateContent' in model.supported_generation_methods:
            print(f"📱 Model: {model.name}")
            print(f"   Display Name: {model.display_name}")
            print(f"   Description: {model.description[:100]}...")
            print(f"   Input Token Limit: {model.input_token_limit}")
            print(f"   Output Token Limit: {model.output_token_limit}")
            print("-" * 50)

explore_gemini_models()
```

## Step 5: Basic Text Generation with Gemini
Demonstrate core text generation functionality using various types of prompts (explanations, code, translations).

```python
def basic_text_generation():
    """Demonstrate basic text generation with Gemini"""
    print("🤖 Basic Text Generation Demo")
    print("=" * 40)

    # Initialize the model
    model = genai.GenerativeModel('gemini-2.0-flash')

    # Sample prompts for different use cases
    prompts = [
        "Explain machine learning in simple terms",
        "Write a Python function to calculate fibonacci numbers",
        "What are the benefits of renewable energy?",
        "Translate 'Hello, how are you?' to French and Spanish"
    ]

    for i, prompt in enumerate(prompts, 1):
        print(f"\n📝 Example {i}: {prompt}")
        print("-" * 30)

        try:
            response = model.generate_content(prompt)
            print(f"🔮 Response: {response.text}")
            print(f"📊 Response length: {len(response.text)} characters")

        except Exception as e:
            print(f"❌ Error: {e}")

        time.sleep(1)  # Rate limiting courtesy

basic_text_generation()
```

## Step 6: Structured API Response Analysis
Examine the complete structure of API responses including metadata, safety ratings, and response properties.

```python
def analyze_api_response():
    """Analyze the structure of Gemini API responses"""
    print("🔬 API Response Structure Analysis")
    print("=" * 45)

    model = genai.GenerativeModel('gemini-2.0-flash')
    prompt = "What is artificial intelligence?"

    try:
        response = model.generate_content(prompt)

        print("📋 Response Object Properties:")
        print(f"   Text: {response.text[:100]}...")
        print(f"   Candidates: {len(response.candidates)}")

        if response.candidates:
            candidate = response.candidates[0]
            print(f"   Finish Reason: {candidate.finish_reason}")
            print(f"   Safety Ratings: {len(candidate.safety_ratings)} ratings")

            # Display safety ratings
            print("\n🛡️  Safety Ratings:")
            for rating in candidate.safety_ratings:
                print(f"{rating.category}: {rating.probability}")

    except Exception as e:
        print(f"❌ Error analyzing response: {e}")

analyze_api_response()
```

## Step 7: Conversation and Chat Functionality
Show how to create multi-turn conversations and maintain chat context across multiple exchanges.

```python
def demonstrate_chat():
    """Demonstrate multi-turn conversation with Gemini"""
    print("💬 Chat Conversation Demo")
    print("=" * 35)

    # Start a chat session
    model = genai.GenerativeModel('gemini-2.0-flash')
    chat = model.start_chat(history=[])

    # Simulate a conversation
    conversation = [
        "Hi, I'm learning about Python programming",
        "Can you explain what a list is?",
        "How is a list different from a tuple?",
        "Can you give me a practical example?"
    ]

    for i, message in enumerate(conversation, 1):
        print(f"\n👤 Human ({i}): {message}")

        try:
            response = chat.send_message(message)
            print(f"🤖 Gemini: {response.text[:150]}...")

        except Exception as e:
            print(f"❌ Error in chat: {e}")

        time.sleep(1)

    # Show chat history
    print(f"\n📚 Total conversation turns: {len(chat.history)}")

demonstrate_chat()
```

## Step 8: Error Handling and Best Practices
Test different scenarios that could cause errors and demonstrate proper exception handling techniques.

```python
def demonstrate_error_handling():
    """Show proper error handling for Gemini API calls"""
    print("⚠️  Error Handling Best Practices")
    print("=" * 40)

    model = genai.GenerativeModel('gemini-2.0-flash')

    # Test cases that might cause errors
    test_cases = [
        ("Normal prompt", "What is the weather like?"),
        ("Very long prompt", "A" * 1000),  # Might hit limits if you increase the number
        ("Empty prompt", ""),
        ("Special characters", "!@#$%^&*()"),
    ]

    for test_name, prompt in test_cases:
        print(f"\n🧪 Testing: {test_name}")
        print(f"   Prompt: {prompt[:50]}...")

        try:
            response = model.generate_content(prompt)
            print(f"   ✅ Success: {len(response.text)} characters generated")

        except Exception as e:
            print(f"   ❌ Error: {type(e).__name__}: {str(e)[:100]}")

demonstrate_error_handling()
```

## Step 9: Utility Functions for API Management
Build a reusable API manager class with retry logic, usage tracking, and production-ready error handling

```python
class GeminiAPIManager:
    """Utility class for managing Gemini API interactions"""

    def __init__(self, model_name='gemini-2.0-flash'):
        self.model = genai.GenerativeModel(model_name)
        self.request_count = 0
        self.total_tokens = 0

    def generate_with_retry(self, prompt: str, max_retries: int = 3):
        """Generate content with retry logic"""
        for attempt in range(max_retries):
            try:
                response = self.model.generate_content(prompt)
                self.request_count += 1
                self.total_tokens += len(prompt.split())  # Rough token estimate
                return response
            except Exception as e:
                if attempt == max_retries - 1:
                    raise e
                time.sleep(2 ** attempt)  # Exponential backoff

    def get_stats(self):
        """Get usage statistics"""
        return {
            "requests_made": self.request_count,
            "estimated_tokens": self.total_tokens
        }

# Test the utility class
print("🛠️  Testing Gemini API Manager")
print("=" * 35)

api_manager = GeminiAPIManager()

test_prompts = [
    "What is Python?",
    "Explain data structures",
    "Benefits of cloud computing"
]

for prompt in test_prompts:
    try:
        response = api_manager.generate_with_retry(prompt)
        print(f"✅ Generated response for: '{prompt[:30]}...'")
    except Exception as e:
        print(f"❌ Failed for prompt: {e}")

print(f"\n📊 Usage Stats: {api_manager.get_stats()}")
```
