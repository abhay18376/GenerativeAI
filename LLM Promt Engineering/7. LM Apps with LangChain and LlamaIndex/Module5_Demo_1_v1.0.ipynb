# Imports

Imports essential modules for building LangChain agents with Groq LLM and DuckDuckGo search tool in Google Colab.

```python
import os
from google.colab import userdata
from langchain.agents import AgentExecutor, create_react_agent, initialize_agent # Import initialize_agent
from langchain.agents import AgentType # Import AgentType
from langchain_groq import ChatGroq
from langchain_core.prompts import ChatPromptTemplate
from langchain_community.tools import DuckDuckGoSearchResults
from langchain.tools import Tool
```

# Set API key

```python
os.environ["GROQ_API_KEY"] = userdata.get("GROQ_API_KEY")
```

# Create tools

Defines a LangChain tools list including a DuckDuckGo web search tool and a calculator tool for math operations.

```python
search = DuckDuckGoSearchResults()
tools = [
    Tool(
        name="Web Search",
        func=search.run,
        description="Current events search"
    ),
    Tool(
        name="Calculator",
        func=lambda x: str(eval(x)),
        description="Math calculations"
    )
]
```

# Build agent

Initializes a ChatGroq language model instance with the Llama 3 8B 8192-token context model at zero temperature for deterministic responses.

```python
llm = ChatGroq(model="llama3-8b-8192", temperature=0)
```

# Use initialize_agent with zero-shot-react-description

Initializes a LangChain agent executor with specified tools and LLM using the Zero-Shot ReAct agent type, enabling verbose output and parsing error handling.

```python
agent_executor = initialize_agent(
    tools, llm, agent=AgentType.ZERO_SHOT_REACT_DESCRIPTION, verbose=True, handle_parsing_errors=True
)
```

# Run query

Executes a LangChain agent to retrieve Meta's current stock price via web search and calculate 15% of the value using a calculator tool, then prints the formatted result. The input can be changed based on the user's preference.

```python
response = agent_executor.invoke({
    "input": "What's Meta's stock price today? Calculate 15% of that value."
})
print(response['output'])
```
