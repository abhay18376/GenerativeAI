# Imports

Imports essential modules for building a document Q&A system using LlamaIndex with Google Gemini LLM and HuggingFace embeddings in Google Colab.

Key Components:
* Environment Setup: os and google.colab.userdata/files for secure API keys and file operations

* LlamaIndex Core: SimpleDirectoryReader (document ingestion), VectorStoreIndex (vector storage), Settings (configuration)

* LLM Integration: GoogleGenAI for Gemini model access

* Embeddings: HuggingFaceEmbedding for open-source text embeddings

This setup enables end-to-end document processing: from file upload → embedding → vector indexing → Gemini-powered querying.

```python
import os
from google.colab import files, userdata
from llama_index.core import SimpleDirectoryReader, VectorStoreIndex, Settings
from llama_index.llms.google_genai import GoogleGenAI
from llama_index.embeddings.huggingface import HuggingFaceEmbedding
```

# 2. Configure API key

Sets the Google API key securely from Colab secrets with error handling and confirmation output.

```python
try:
    os.environ["GOOGLE_API_KEY"] = userdata.get("GEMINI_NEW")
    print("✅ Google API key configured successfully")
except Exception as e:
    print(f"❌ Error setting API key: {str(e)}")
    raise
```

# 3. Set models

Configures the LlamaIndex settings to use Google's Gemini 1.5 Flash model as the LLM and the BAAI/bge-small-en-v1.5 HuggingFace model for embeddings, with a confirmation print statement.

```python
Settings.llm = GoogleGenAI(model="gemini-1.5-flash-latest")
Settings.embed_model = HuggingFaceEmbedding(model_name="BAAI/bge-small-en-v1.5")
print("⚙️ Models configured: Gemini 1.5 Flash + BGE embeddings")
```

# 4. Upload documents

Uploads files via Google Colab, ensures a 'docs' directory exists, saves each uploaded file there, and prints upload confirmation.

```python
uploaded = files.upload()
os.makedirs("docs", exist_ok=True)
for filename in uploaded.keys():
    with open(f"docs/{filename}", "wb") as f:
        f.write(uploaded[filename])
    print(f"📄 Uploaded: {filename}")
```

# 5. Build index

Attempts to load documents from the 'docs' directory and build a vector index using LlamaIndex, with success/failure logging and error propagation.

```python
try:
    documents = SimpleDirectoryReader("docs").load_data()
    index = VectorStoreIndex.from_documents(documents)
    print("✅ Index built successfully")
except Exception as e:
    print(f"❌ Indexing failed: {str(e)}")
    raise
```

# 6. Query with enhanced settings

Configures a LlamaIndex query engine to retrieve top 5 relevant document chunks and perform hierarchical summarization, then executes a summarization query while capturing response metadata.

Key Features:
* Enhanced Retrieval (similarity_top_k=5):

    * Fetches 5 most relevant text chunks for richer context

* Advanced Summarization (response_mode="tree_summarize"):

   * Uses recursive tree-based summarization for coherent document distillation

* Metadata Capture:

  Prints both the summary response and detailed metadata

```python
query_engine = index.as_query_engine(
    similarity_top_k=5,  # Retrieve top 5 relevant chunks
    response_mode="tree_summarize"  # Better for summarization
)
response = query_engine.query("Summarize key points from the document")
print("\n💬 Response:", response)
print("\n🔍 Response Metadata:", response.metadata)
```
