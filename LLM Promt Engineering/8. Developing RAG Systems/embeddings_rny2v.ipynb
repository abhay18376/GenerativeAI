```python
from sentence_transformers import SentenceTransformer, util
sentences = ["This is an example sentence", "Each sentence is converted"]
model = SentenceTransformer('sentence-transformers/all-MiniLM-L6-v2')
embeddings = model.encode(sentences)
print(embeddings)
```

```python
# Example corpus (like search engine documents)
corpus = [
    "How to learn Python programming effectively", # 0
    "Best restaurants near Hyderabad", # 1
    "Latest trends in artificial intelligence", # 2
    "Tips for improving machine learning models", # 3
    "Weather forecast for Bangalore tomorrow", # 4
    "Beginner's guide to deep learning and neural networks",
]
query = "refer me some books on AI"

# Encode both
query_embedding = model.encode(query, convert_to_tensor=True)
corpus_embeddings = model.encode(corpus, convert_to_tensor=True)

# Compute cosine similarity
cosine_scores = util.cos_sim(query_embedding, corpus_embeddings)

# Rank results
top_results = cosine_scores[0].argsort(descending=True)
```

```python
import time
start_time = time.time()
query = "give me climate condition" # 5

query_embedding = model.encode(query, convert_to_tensor=True) #tensor - array of data -(1,384)
cosine_scores = util.cos_sim(query_embedding, corpus_embeddings) # (1,384).(384,6) = (1,6)
#
# Rank results
top_results = cosine_scores[0].argsort(descending=True)
end_time = time.time()

print(top_results)
print(f"Time taken: {end_time - start_time} seconds")
```
