```python
import os
path ="/content/Rag_documents"

# read the content
def read_content(path):
  docs =[]
  for file in os.listdir(path):
    print(file)
    # ext = os.path.splitext(path)[1].lower()
    if file.endswith('.txt'):
      loader = TextLoader("/content/Rag_documents/"+str(file))
    elif file.endswith('.pdf'):
      loader = PyPDFLoader("/content/Rag_documents/"+str(file))
    elif file.endswith('.docx'):
      loader = Docx2txtLoader("/content/Rag_documents/"+str(file))
    else:
      # print(file)
      print('File type not supported')
      continue
    docs.extend(loader.load())
  return(docs)

# docs_list = read_content(path)

text_splitter = RecursiveCharacterTextSplitter(
                chunk_size=500,
                chunk_overlap=50,
                separators=["\n\n", "\n", " ", ""],
                length_function=len,
                keep_separator=False
            )
split_docs = text_splitter.split_documents(docs)
```
